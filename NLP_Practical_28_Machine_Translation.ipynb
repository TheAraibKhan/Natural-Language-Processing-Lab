```python
# Practical 28: Machine Translation
# Objective: Translate text from English to French using a pretrained transformer model.

# Install required library
!pip -q install transformers sentencepiece

# Import required libraries
from transformers import MarianMTModel, MarianTokenizer

# Load pretrained English-to-French translation model
model_name = "Helsinki-NLP/opus-mt-en-fr"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# Input English text
text = "Artificial intelligence is changing the way people learn and work."

# Tokenize the input text
inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True)

# Generate translation
translated_tokens = model.generate(
    **inputs,
    max_new_tokens=100
)

# Decode translated tokens
translation = tokenizer.decode(
    translated_tokens[0],
    skip_special_tokens=True
)

# Display results
print("Original Text (English):")
print(text)

print("\nTranslated Text (French):")
print(translation)
```
