```python
# ============================================================
# NATURAL LANGUAGE PROCESSING LABORATORY
#
# Practical 17: Topic Modeling using Latent Dirichlet Allocation
# (LDA)
#
# Name: Araib Khan
# Course: B.Tech CSE (Artificial Intelligence)
# Subject: Natural Language Processing Laboratory
# Practical No.: 17
# CO: CO2
#
# Objective:
# To discover hidden topics in a collection of text documents
# using Latent Dirichlet Allocation (LDA).
# ============================================================

# Step 1: Install required libraries

!pip install -q scikit-learn pandas matplotlib

# Step 2: Import libraries

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

print("Libraries imported successfully.")

# Step 3: Create a sample text corpus

documents = [
    # Technology documents
    "Artificial intelligence and machine learning are transforming technology.",
    "Deep learning algorithms are used in artificial intelligence applications.",
    "Python programming is useful for software development and data science.",
    "Computer networks connect devices and enable digital communication.",
    "Machine learning models analyze data and make intelligent predictions.",

    # Sports documents
    "Football players practice regularly before the championship tournament.",
    "Cricket players improve their batting bowling and fielding skills.",
    "Basketball teams compete in tournaments and professional leagues.",
    "The athlete won a gold medal in the international sports competition.",
    "Football and cricket are popular sports around the world.",

    # Food documents
    "Cooking delicious food requires fresh vegetables and spices.",
    "Italian pizza and pasta are popular food dishes.",
    "Healthy eating includes fruits vegetables and nutritious meals.",
    "Chefs prepare delicious recipes using different cooking methods.",
    "Fresh fruits and vegetables are important for a healthy diet."
]

print("\nSample Text Corpus:")
for i, document in enumerate(documents, 1):
    print(f"Document {i}: {document}")

# Step 4: Convert text into a document-term matrix

vectorizer = CountVectorizer(stop_words="english")

document_term_matrix = vectorizer.fit_transform(documents)

print("\nDocument-Term Matrix:")
print(document_term_matrix.toarray())

print("\nDocument-Term Matrix Shape:")
print(document_term_matrix.shape)

# Step 5: Display vocabulary

feature_names = vectorizer.get_feature_names_out()

print("\nVocabulary:")
print(feature_names)

print("\nVocabulary Size:", len(feature_names))

# Step 6: Display document-term matrix

dtm_table = pd.DataFrame(
    document_term_matrix.toarray(),
    columns=feature_names
)

print("\nDocument-Term Matrix Table:")
display(dtm_table)

# Step 7: Initialize the LDA model

lda_model = LatentDirichletAllocation(
    n_components=3,
    random_state=42,
    max_iter=30,
    learning_method="batch"
)

# Step 8: Train the LDA model

lda_model.fit(document_term_matrix)

print("\nLDA Model Trained Successfully.")

# Step 9: Extract topics and important words

num_words = 10

print("\nDiscovered Topics:")
print("=" * 60)

for topic_index, topic in enumerate(lda_model.components_):
    top_word_indices = topic.argsort()[-num_words:][::-1]

    top_words = [
        feature_names[index]
        for index in top_word_indices
    ]

    print(f"\nTopic {topic_index + 1}:")
    print(", ".join(top_words))

# Step 10: Display topic-word weights

print("\nTopic-Word Weights:")

for topic_index, topic in enumerate(lda_model.components_):
    top_word_indices = topic.argsort()[-10:][::-1]

    topic_data = pd.DataFrame({
        "Word": feature_names[top_word_indices],
        "Weight": topic[top_word_indices]
    })

    print(f"\nTopic {topic_index + 1}:")
    display(topic_data)

# Step 11: Calculate document-topic distributions

document_topic_matrix = lda_model.transform(
    document_term_matrix
)

topic_columns = [
    f"Topic {i + 1}"
    for i in range(lda_model.n_components)
]

topic_table = pd.DataFrame(
    document_topic_matrix,
    columns=topic_columns
)

topic_table.index = [
    f"Document {i + 1}"
    for i in range(len(documents))
]

print("\nDocument-Topic Distribution:")
display(topic_table.round(4))

# Step 12: Identify dominant topic for each document

dominant_topics = document_topic_matrix.argmax(axis=1) + 1

document_results = pd.DataFrame({
    "Document": documents,
    "Dominant Topic": dominant_topics,
    "Topic Probability": document_topic_matrix.max(axis=1)
})

print("\nDominant Topic for Each Document:")
display(document_results.round(4))

# Step 13: Visualize topic distribution

topic_counts = pd.Series(dominant_topics).value_counts().sort_index()

topic_counts.index = [
    f"Topic {i}"
    for i in topic_counts.index
]

plt.figure(figsize=(8, 5))

topic_counts.plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Document Distribution Across LDA Topics")
plt.xlabel("Topics")
plt.ylabel("Number of Documents")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Step 14: Test the model on a new document

new_documents = [
    "Artificial intelligence is used in modern computer applications.",
    "The cricket team performed well in the tournament.",
    "Fresh vegetables and fruits are important for healthy food."
]

new_document_matrix = vectorizer.transform(new_documents)

new_topic_distribution = lda_model.transform(
    new_document_matrix
)

print("\nTopic Prediction for New Documents:")
print("=" * 60)

for i, document in enumerate(new_documents):
    predicted_topic = (
        new_topic_distribution[i].argmax() + 1
    )

    print(f"\nDocument: {document}")
    print(f"Predicted Dominant Topic: Topic {predicted_topic}")
    print(
        "Topic Distribution:",
        new_topic_distribution[i].round(4)
    )

# Step 15: Display model information

print("\nLDA Model Information:")
print("-" * 40)

print("Number of Topics:", lda_model.n_components)
print("Vocabulary Size:", len(feature_names))
print("Number of Documents:", len(documents))
print("Maximum Iterations:", lda_model.max_iter)

# Step 16: Result and Conclusion

print("\n" + "=" * 60)
print("RESULT AND CONCLUSION")
print("=" * 60)

print("""
Result:
Latent Dirichlet Allocation (LDA) was successfully implemented
on the given text corpus.

The document-term matrix was generated, and the LDA model
discovered three latent topics based on word distributions.

The important words associated with each topic were extracted,
and document-topic distributions were calculated.

Conclusion:
LDA is an unsupervised topic modeling technique used to discover
hidden thematic structures in a collection of documents.

It represents documents as mixtures of topics and topics as
distributions over words. It is useful for document analysis,
information retrieval, content organization, and text analytics.
""")
```
