```python
import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk import pos_tag
from nltk import RegexpParser
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
sentence = "The intelligent student is developing a machine learning project."

print("Input Sentence:")
print(sentence)
```

```python
words = word_tokenize(sentence)
pos_tags = pos_tag(words)

print("POS Tags:")
for word, tag in pos_tags:
    print(f"{word:15} -> {tag}")
```

```python
grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>+}
    VP: {<MD>?<VB.*><RB>*}
"""
```

```python
chunk_parser = RegexpParser(grammar)

chunk_tree = chunk_parser.parse(pos_tags)

print("Regex Chunking Result:")
print(chunk_tree)
```

```python
print("Chunk Tree:\n")
chunk_tree.pretty_print()
```

```python
print("Extracted Chunks:\n")

for subtree in chunk_tree.subtrees():
    if subtree.label() in ["NP", "VP"]:
        phrase = " ".join(word for word, tag in subtree.leaves())
        print(f"{subtree.label()} -> {phrase}")
```

```python
doc = nlp(sentence)

print("spaCy Tokens and Dependencies:\n")

for token in doc:
    print(
        f"{token.text:15} "
        f"POS={token.pos_:10} "
        f"DEP={token.dep_:12} "
        f"HEAD={token.head.text}"
    )
```

```python
print("Noun Phrases using spaCy:\n")

for chunk in doc.noun_chunks:
    print(f"Noun Phrase: {chunk.text}")
```

```python
print("Dependency Relationships:\n")

for token in doc:
    print(
        f"{token.text:15} "
        f"--> {token.dep_:12} "
        f"--> {token.head.text}"
    )
```

```python
print("Comparison")
print("-" * 50)

print("\nNLTK Regex Chunking:")
for subtree in chunk_tree.subtrees():
    if subtree.label() in ["NP", "VP"]:
        phrase = " ".join(word for word, tag in subtree.leaves())
        print(f"{subtree.label()} -> {phrase}")

print("\nspaCy Noun Chunking:")
for chunk in doc.noun_chunks:
    print(f"NP -> {chunk.text}")
```
