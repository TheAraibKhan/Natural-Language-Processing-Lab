```python
import gensim.downloader as api
import numpy as np

print("Libraries imported successfully.")
```

```python
print("Loading pre-trained GloVe embeddings...")

model = api.load("glove-wiki-gigaword-50")

print("GloVe embeddings loaded successfully.")
print("Vocabulary Size:", len(model))
print("Vector Dimension:", model.vector_size)
```

```python
document1 = "machine learning is a useful technology for data analysis"

document2 = "machine learning is useful for analyzing data"

document3 = "football players practice on the field for the game"

print("Document 1:")
print(document1)

print("\nDocument 2:")
print(document2)

print("\nDocument 3:")
print(document3)
```

```python
tokens1 = document1.lower().split()
tokens2 = document2.lower().split()
tokens3 = document3.lower().split()

print("Document 1 Tokens:")
print(tokens1)

print("\nDocument 2 Tokens:")
print(tokens2)

print("\nDocument 3 Tokens:")
print(tokens3)
```

```python
def check_vocabulary(tokens, model):
    missing_words = [word for word in tokens if word not in model]
    return missing_words


print("Vocabulary Check")

missing1 = check_vocabulary(tokens1, model)
missing2 = check_vocabulary(tokens2, model)
missing3 = check_vocabulary(tokens3, model)

print("Document 1 missing words:", missing1)
print("Document 2 missing words:", missing2)
print("Document 3 missing words:", missing3)
```

```python
wmd_1_2 = model.wmdistance(tokens1, tokens2)

print("Word Mover's Distance")
print("----------------------")
print(f"Document 1 vs Document 2: {wmd_1_2:.4f}")
```

```python
wmd_1_3 = model.wmdistance(tokens1, tokens3)

print("Word Mover's Distance")
print("----------------------")
print(f"Document 1 vs Document 3: {wmd_1_3:.4f}")
```

```python
wmd_2_3 = model.wmdistance(tokens2, tokens3)

print("Word Mover's Distance")
print("----------------------")
print(f"Document 2 vs Document 3: {wmd_2_3:.4f}")
```

```python
print("Pairwise Word Mover's Distance")
print("=" * 50)

print(f"Document 1 vs Document 2 : {wmd_1_2:.4f}")
print(f"Document 1 vs Document 3 : {wmd_1_3:.4f}")
print(f"Document 2 vs Document 3 : {wmd_2_3:.4f}")
```

```python
documents = [tokens1, tokens2, tokens3]

distance_matrix = np.zeros((3, 3))

for i in range(3):
    for j in range(3):
        if i == j:
            distance_matrix[i][j] = 0.0
        else:
            distance_matrix[i][j] = model.wmdistance(
                documents[i],
                documents[j]
            )

print("WMD Distance Matrix:")
print(distance_matrix.round(4))
```

```python
import pandas as pd

distance_table = pd.DataFrame(
    distance_matrix,
    index=["Document 1", "Document 2", "Document 3"],
    columns=["Document 1", "Document 2", "Document 3"]
)

print("Word Mover's Distance Matrix:")
display(distance_table.round(4))
```

```python
pairs = {
    "Document 1 vs Document 2": wmd_1_2,
    "Document 1 vs Document 3": wmd_1_3,
    "Document 2 vs Document 3": wmd_2_3
}

closest_pair = min(pairs, key=pairs.get)

print("Closest Document Pair:")
print(closest_pair)

print(f"\nWMD Score: {pairs[closest_pair]:.4f}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text_documents = [
    document1,
    document2,
    document3
]

vectorizer = TfidfVectorizer()
tfidf_vectors = vectorizer.fit_transform(text_documents)

cosine_matrix = cosine_similarity(tfidf_vectors)

print("Cosine Similarity Matrix:")
print(cosine_matrix.round(4))
```

```python
print("Comparison of Similarity Measures")
print("=" * 55)

print(f"WMD - Document 1 vs Document 2 : {wmd_1_2:.4f}")
print(
    f"Cosine Similarity - Document 1 vs Document 2 : "
    f"{cosine_matrix[0][1]:.4f}"
)

print("\nNote:")
print("WMD is a distance measure, so lower values indicate greater similarity.")
print("Cosine similarity is a similarity measure, so higher values indicate greater similarity.")
```

```python
# Word Mover's Distance uses word embeddings to measure how much
# "movement" is required to transform the words of one document into
# the words of another document.
#
# Lower WMD:
# Documents are semantically closer.
#
# Higher WMD:
# Documents are semantically farther apart.
```

```python
# Result:
# The Word Mover's Distance between the given documents was successfully
# calculated using pre-trained GloVe word embeddings.
#
# Pairwise distances and a complete WMD distance matrix were generated.
# The documents with more related semantic content produced lower
# WMD values than the unrelated document pair.
#
# Conclusion:
# Word Mover's Distance is an embedding-based text similarity technique
# that measures semantic distance between documents. Unlike simple
# word-overlap methods, WMD can use relationships between word vectors
# to capture semantic similarity even when documents do not contain
# exactly the same words.
```
