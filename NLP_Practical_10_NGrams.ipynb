```python
import nltk
from nltk.util import ngrams
from collections import Counter

print("NLTK imported successfully.")
```

```python
corpus = """
Natural language processing is a branch of artificial intelligence.
Natural language processing helps computers understand human language.
Machine learning is widely used in natural language processing.
"""

print("Corpus:")
print(corpus)
```

```python
tokens = corpus.lower().split()

print("Tokens:")
print(tokens)
```

```python
unigrams = list(ngrams(tokens, 1))

print("Unigrams:")
print(unigrams)
```

```python
bigrams = list(ngrams(tokens, 2))

print("Bigrams:")
print(bigrams)
```

```python
trigrams = list(ngrams(tokens, 3))

print("Trigrams:")
print(trigrams)
```

```python
print("Unigram Sequences:\n")

for unigram in unigrams:
    print(" ".join(unigram))
```

```python
print("Bigram Sequences:\n")

for bigram in bigrams:
    print(" ".join(bigram))
```

```python
print("Trigram Sequences:\n")

for trigram in trigrams:
    print(" ".join(trigram))
```

```python
unigram_counts = Counter(unigrams)
bigram_counts = Counter(bigrams)
trigram_counts = Counter(trigrams)

print("Most Common Unigrams:")
for gram, count in unigram_counts.most_common(10):
    print(" ".join(gram), "->", count)

print("\nMost Common Bigrams:")
for gram, count in bigram_counts.most_common(10):
    print(" ".join(gram), "->", count)

print("\nMost Common Trigrams:")
for gram, count in trigram_counts.most_common(10):
    print(" ".join(gram), "->", count)
```

```python
print("N-Gram Count Comparison")
print("-" * 40)

print("Total Unigrams :", len(unigrams))
print("Total Bigrams  :", len(bigrams))
print("Total Trigrams :", len(trigrams))
```

```python
def generate_ngrams(tokens, n):
    return list(ngrams(tokens, n))


print("Generated N-Grams:\n")

for n in [1, 2, 3]:
    result = generate_ngrams(tokens, n)

    print(f"{n}-Gram:")

    for gram in result[:10]:
        print(" ".join(gram))

    print()
```

```python
print("Example of N-Grams")
print("-" * 40)

print("Unigram :", "natural")
print("Bigram  :", "natural language")
print("Trigram :", "natural language processing")
```

```python
# Result:
# Unigrams, bigrams, and trigrams were successfully generated from the
# given text corpus using NLTK.
#
# The frequencies of the generated N-Grams were also calculated and
# compared.
#
# Conclusion:
# N-Gram models represent local word sequences in a corpus. Unigrams
# represent individual words, bigrams represent two-word sequences,
# and trigrams represent three-word sequences. N-Grams are useful for
# modeling local context and word relationships in Natural Language
# Processing applications.
```
